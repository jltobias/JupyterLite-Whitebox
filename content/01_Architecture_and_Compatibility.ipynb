# Architecture and JupyterLite compatibility

The PyPI `whitebox` package is a Python interface to a standalone **WhiteboxTools native executable**. Its `WhiteboxTools` class launches that executable using Python subprocesses. JupyterLite runs CPython-compatible Python compiled to WebAssembly (Pyodide) inside the browser, where native OS executables cannot be launched.

That gives us a useful split: **inspect and learn the Python API in JupyterLite; execute geoprocessing in Binder/local Jupyter/Colab.**


````python
%pip install -q whitebox
import inspect
from whitebox.whitebox_tools import WhiteboxTools
print(inspect.getdoc(WhiteboxTools).splitlines()[0])
print('Public wrapper methods:', len([n for n in dir(WhiteboxTools) if not n.startswith('_')]))
````

````python
import sys
if 'pyodide' in sys.modules:
    print('Browser kernel: do not instantiate WhiteboxTools here.')
else:
    import whitebox
    wbt = whitebox.WhiteboxTools()
    print(wbt.version())
````

## Native-kernel pattern
In local Jupyter/Binder, the normal pattern is:
```python
import whitebox
wbt = whitebox.WhiteboxTools()
wbt.set_working_dir('/path/to/data')
wbt.slope('DEM.tif', 'slope.tif')
```
