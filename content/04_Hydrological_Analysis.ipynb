````python
import numpy as np, matplotlib.pyplot as plt
y,x=np.mgrid[-2:2:120j,-2:2:120j]
dem=100 + 8*x + 5*y + 15*np.exp(-(x*x+y*y)*2) - 9*np.exp(-((x-.5)**2+(y+.4)**2)*12)
plt.imshow(dem)
plt.title('Synthetic DEM for hydrology concepts')
plt.colorbar(label='elevation');
````
