# Searchable Whitebox Python API catalog
This notebook turns the installed `whitebox` wrapper into a lightweight browser-searchable catalog. It does **not** instantiate the native engine. Change `query` to explore tools by method name or docstring.


```python
%pip install -q whitebox
from _helpers import api_search
query=['slope']
rows=api_search(query,100)
print('matches:',len(rows))
for name,sig,summary in rows:
    print(f'\n{name}{sig}\n  {summary}')
```

```python
# Browse all public wrapper methods
all_rows=api_search([],1000)
print('Public callable wrapper methods discovered:',len(all_rows))
print('\n'.join(name for name,_,_ in all_rows[:120]))
```

```python
# Inspect the full docstring and signature for a method without running it
import inspect
from whitebox.whitebox_tools import WhiteboxTools
method_name='slope'
method=getattr(WhiteboxTools,method_name)
print(method_name, inspect.signature(method))
print(inspect.getdoc(method))
```
