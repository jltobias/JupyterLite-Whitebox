````python
import numpy as np, matplotlib.pyplot as plt
y,x=np.mgrid[-3:3:160j,-3:3:160j]
dem=300 + 55*np.exp(-(x*x+y*y)/2) + 12*np.sin(2*x)*np.cos(1.5*y)
gy,gx=np.gradient(dem)
slope=np.degrees(np.arctan(np.hypot(gx,gy)))
plt.imshow(slope)
plt.title('Slope concept from synthetic DEM')
plt.colorbar(label='degrees (illustrative)');
````
