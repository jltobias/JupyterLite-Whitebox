```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(7)
x=rng.uniform(0,100,1500); y=rng.uniform(0,100,1500)
z=15 + .08*x + .03*y + 8*np.exp(-((x-55)**2+(y-50)**2)/500) + rng.normal(0,.35,1500)
plt.scatter(x,y,c=z,s=5)
plt.title('Synthetic point cloud concept')
plt.xlabel('x'); plt.ylabel('y');
```
