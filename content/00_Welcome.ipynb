## Whitebox overview

Whitebox Geospatial Analysis Tools is an open-source, cross-platform GIS and remote-sensing software project associated with the University of Guelph’s Centre for Hydrogeomatics. It is designed for research and education and supports raster, vector, and LiDAR workflows, with scripting interfaces that include Python. The tool families represented in this book include GIS operations, image processing, hydrology, terrain analysis, stream-network analysis, and LiDAR processing.

Whitebox is especially useful for transparent, reproducible geospatial analysis: a workflow can be described as a sequence of named tools with explicit inputs and outputs. In this collection, the `whitebox` Python package exposes the tool interface, while JupyterLite provides browser-based API exploration. Actual processing of GeoTIFF, GeoPackage, and LAS/LAZ files is demonstrated as native-kernel code because the browser runtime cannot launch the native WhiteboxTools executable.

For additional background, see [Whitebox Geospatial Analysis Tools on Wikipedia](https://en.wikipedia.org/wiki/Whitebox_Geospatial_Analysis_Tools) and the [WhiteboxTools project](https://github.com/jblindsay/whitebox-tools).
