```python
import numpy as np
rng=np.random.default_rng(10)
a=rng.normal(100,15,(50,50))
print('mean',a.mean())
print('std',a.std())
z=(a-a.mean())/a.std()
print('normalized mean/std',round(float(z.mean()),6),round(float(z.std()),6))
```
